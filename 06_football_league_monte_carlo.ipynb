```r
count_points <- function(outcomes, table0, games) {
    tab <- table0
  
    for(i in 1:nrow(games)) {
        home <- games$h[i]
        away <- games$a[i]
        res  <- outcomes[i]
    
        if(res == 0) {                # wygrana gospodarzy
            tab$points[tab$labels == home] <- tab$points[tab$labels == home] + 3
        } else if(res == 1) {         # remis
            tab$points[tab$labels == home] <- tab$points[tab$labels == home] + 1
            tab$points[tab$labels == away] <- tab$points[tab$labels == away] + 1
        } else if(res == 2) {         # wygrana gości
            tab$points[tab$labels == away] <- tab$points[tab$labels == away] + 3
        }
    }
  
    tab <- tab[order(-tab$points, tab$labels), ]
    rownames(tab) <- NULL
    tab
}
```
