```r
library(animation)

Lnew<-matrix(nrow=10,ncol=10,0)
L<-matrix(nrow=10,ncol=10,0)
#warunki brzegowe
L[,1]<-rep(0,10)
L[,10]<-rep(0,10)
L[1,]<-rep(0,10)
L[10,]<-rep(1,10)
#chcemy mieć warunki brzegowe też w nowym kroku
Lnew<-L
#wypisanie stanu pola w kroku 0
L
biggest_change = 1
for (k in 1:100)
{
#pętla po wierszach (i) i kolumnach (j)
    print("iteracja")
    print(k)
    print("zmiana")
    print(biggest_change)
    
    if(biggest_change < 0.001)
        {
        break
        }
    biggest_change = 0
for (i in 2:9){
for (j in 2:9){
Lnew[i,j]<-0.25*(L[i-1,j]+L[i+1,j]+L[i,j-1]+L[i,j+1])
    change = Lnew[i,j] - L[i,j]
    if (change >biggest_change)
        {
        biggest_change = change
        }
 }
}
#przejście o krok do przodu w iteracji
L<-Lnew
} #po k
#wypisanie wyniku
L
Limg <- apply(L, 2, rev)
image(t(Limg))
```
