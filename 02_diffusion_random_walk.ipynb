```python
import numpy as np
import matplotlib.pyplot as plt
import random
import matplotlib.animation as animation
from IPython.display import HTML

fig = plt.figure("Test")
ims = []
N = 100

Lnew = np.zeros((N, N))
L = np.zeros((N, N))
L[:, 0] = L[:, N-1] = L[0, :] = L[N-1, :] = 0
Lnew = np.copy(L)
pos = [50, 50]
L[pos[0], pos[1]] = 1

for k in range(1,1001):
    Lnew[1:N-1, 1:N-1] = 0.25 * (L[:-2, 1:N-1] + L[2:, 1:N-1] + L[1:N-1, :-2] + L[1:N-1, 2:])
    Lnew[pos[0],pos[1]]=1
    L = np.copy(Lnew)
    if k%10==0:
        ims.append([plt.imshow(L,cmap='hot'),plt.text(10, 10, str(k), color='white', fontsize=12)])
    while True:
        pom = [pos[0] + random.randint(-1, 1), pos[1] + random.randint(-1, 1)]
        if 0 < pom[0] < N-1 and 0 < pom[1] < N-1:
            break
    pos = pom
    L[pos[0], pos[1]] = 1

ani = animation.ArtistAnimation(fig, ims, interval=100, blit=True, repeat = False)
plt.close()
from IPython.display import HTML
HTML(ani.to_html5_video())
```
